```python
# Erstellung einer Funktion um den SOH der Zellen zu berechnen
def SOH_Berechnen_C2_C10(Zellen):
    # Ermittlung der Zelle die den kürzesten Zyklenzähler hat, damit alle anderen Zellen auf diese länge gekürzt werden können
    Kürzeste_Zelle = min(z["Neuer_Cyc-Count"].max() for z in Zellen)
    # Erstellung einer Liste für die jeweilgen Zellen
    Zellen_SOH = []
    # Beginn einer Schleife, um den individuellen SOH jeder betrachteten Zelle zu berechnen
    for i, Zelle in enumerate(Zellen):
        # Erstellung einer Kopie der jeweiligen Zelle, um die Fehlermedlung von Pandas zu vermeiden
        Zelle = Zelle.copy()
        # Kürzen der Zelle auf den kürzesten Zyklenzählwert und Sortieren nach Zyklenzählwert
        Zelle = Zelle[Zelle["Neuer_Cyc-Count"] <= Kürzeste_Zelle].sort_values(by="Neuer_Cyc-Count").reset_index(drop=True)
        # Der SOH wird an dem ersten gemessenen Wert der C/2 Zyklen bemessen
        # Dazu wird dieser als Anfangskapazität definiert
        Anfangs_Kapazität = Zelle['Spezifische_Kapazität_mAh_g'].iloc[0]
        # Berechnung des aktuellen SOH der Mittelwerte in % indem der aktuelle Wert der spezifischen Kapazität durch die Anfangskapazität geteilt und mit 100 multipliziert wird
        Zelle['SOH_Prozent'] = (Zelle['Spezifische_Kapazität_mAh_g'] / Anfangs_Kapazität) * 100
        # Anhängen der berechneten SOH-Werte der jeweiligen Zelle an die Liste
        Zellen_SOH.append(Zelle)
    # Zusammenführen aller Zellen in einen DataFrame
    SOH_Gesamt = pd.concat(Zellen_SOH, ignore_index=True)
    # Berechnung des Mittelwertes und der zugehörigen Standardabweichung aller angegebenen Zellen
    SOH_Mittelwert = SOH_Gesamt.groupby('Neuer_Cyc-Count')['SOH_Prozent'].mean().reset_index()
    SOH_Standardabweichung = SOH_Gesamt.groupby('Neuer_Cyc-Count')['SOH_Prozent'].std().reset_index()
    # Rückgabe des Mittelwertes sowie der Standardabweichung
    return SOH_Mittelwert, SOH_Standardabweichung


# Berechnung des SOH für die Referenzzellen C/2-Zyklen
Zellen_Ref_1h_C2_Mittlere_Spezifische_Kapazität_SOH, Zellen_Ref_1h_C2_Std_SOH = SOH_Berechnen_C2_C10([Zelle_Ref_1h_VZ_1_C2_Lücke_Zyklen, Zelle_Ref_1h_VZ_2_C2_Lücke_Zyklen, Zelle_Ref_1h_VZ_3_C2_Lücke_Zyklen, Zelle_Ref_1h_VZ_4_C2_Lücke_Zyklen])
# Berechnung des SOH für die Referenzzellen C/10-Zyklen
Zellen_Ref_1h_C10_Mittlere_Spezifische_Kapazität_SOH, Zellen_Ref_1h_C10_Std_SOH = SOH_Berechnen_C2_C10([Zelle_Ref_1h_VZ_1_C10_Lücke_Zyklen, Zelle_Ref_1h_VZ_2_C10_Lücke_Zyklen, Zelle_Ref_1h_VZ_3_C10_Lücke_Zyklen, Zelle_Ref_1h_VZ_4_C10_Lücke_Zyklen])

# Berechnung des SOH für die Zellen mit 5 µm Li-Folie C/2-Zyklen
Zellen_5um_1h_C2_Mittlere_Spezifische_Kapazität_SOH,  Zellen_5um_1h_C2_Std_SOH = SOH_Berechnen_C2_C10([Zelle_5um_1h_VZ_1_C2_Lücke_Zyklen, Zelle_5um_1h_VZ_2_C2_Lücke_Zyklen, Zelle_5um_1h_VZ_3_C2_Lücke_Zyklen, Zelle_5um_1h_VZ_4_C2_Lücke_Zyklen])
# Berechnung des SOH für die Zellen mit 5 µm Li-Folie C/10-Zyklen
Zellen_5um_1h_C10_Mittlere_Spezifische_Kapazität_SOH,  Zellen_5um_1h_C10_Std_SOH = SOH_Berechnen_C2_C10([Zelle_5um_1h_VZ_1_C10_Lücke_Zyklen, Zelle_5um_1h_VZ_2_C10_Lücke_Zyklen, Zelle_5um_1h_VZ_3_C10_Lücke_Zyklen, Zelle_5um_1h_VZ_4_C10_Lücke_Zyklen])

# Berechnung des SOH für die Zellen mit 5 µm Li-Folie C/2-Zyklen
Zellen_5um_24h_C2_Mittlere_Spezifische_Kapazität_SOH,  Zellen_5um_24h_C2_Std_SOH = SOH_Berechnen_C2_C10([Zelle_5um_24h_VZ_1_C2_Lücke_Zyklen, Zelle_5um_24h_VZ_2_C2_Lücke_Zyklen, Zelle_5um_24h_VZ_3_C2_Lücke_Zyklen])
# Berechnung des SOH für die Zellen mit 5 µm Li-Folie C/10-Zyklen
Zellen_5um_24h_C10_Mittlere_Spezifische_Kapazität_SOH,  Zellen_5um_24h_C10_Std_SOH = SOH_Berechnen_C2_C10([Zelle_5um_24h_VZ_1_C10_Lücke_Zyklen, Zelle_5um_24h_VZ_2_C10_Lücke_Zyklen, Zelle_5um_24h_VZ_3_C10_Lücke_Zyklen])

# Berechnung des SOH für die Zellen mit 5 µm Li-Folie C/2-Zyklen
Zellen_5um_48h_C2_Mittlere_Spezifische_Kapazität_SOH,  Zellen_5um_48h_C2_Std_SOH = SOH_Berechnen_C2_C10([Zelle_5um_48h_VZ_1_C2_Lücke_Zyklen, Zelle_5um_48h_VZ_2_C2_Lücke_Zyklen, Zelle_5um_48h_VZ_3_C2_Lücke_Zyklen, Zelle_5um_48h_VZ_4_C2_Lücke_Zyklen])
# Berechnung des SOH für die Zellen
Zellen_5um_48h_C10_Mittlere_Spezifische_Kapazität_SOH,  Zellen_5um_48h_C10_Std_SOH = SOH_Berechnen_C2_C10([Zelle_5um_48h_VZ_1_C10_Lücke_Zyklen, Zelle_5um_48h_VZ_2_C10_Lücke_Zyklen, Zelle_5um_48h_VZ_3_C10_Lücke_Zyklen, Zelle_5um_48h_VZ_4_C10_Lücke_Zyklen])

# Berechnung des SOH für die Zellen mit 5 µm Li-Folie C/2-Zyklen
Zellen_5um_168h_C2_Mittlere_Spezifische_Kapazität_SOH,  Zellen_5um_168h_C2_Std_SOH = SOH_Berechnen_C2_C10([Zelle_5um_168h_VZ_1_C2_Lücke_Zyklen, Zelle_5um_168h_VZ_2_C2_Lücke_Zyklen, Zelle_5um_168h_VZ_3_C2_Lücke_Zyklen, Zelle_5um_168h_VZ_4_C2_Lücke_Zyklen, Zelle_5um_168h_VZ_5_C2_Lücke_Zyklen, Zelle_5um_168h_VZ_6_C2_Lücke_Zyklen])
# Berechnung des SOH für die Zellen mit 5 µm Li-Folie C/10-Zyklen
Zellen_5um_168h_C10_Mittlere_Spezifische_Kapazität_SOH,  Zellen_5um_168h_C10_Std_SOH = SOH_Berechnen_C2_C10([Zelle_5um_168h_VZ_1_C10_Lücke_Zyklen, Zelle_5um_168h_VZ_2_C10_Lücke_Zyklen, Zelle_5um_168h_VZ_3_C10_Lücke_Zyklen, Zelle_5um_168h_VZ_4_C10_Lücke_Zyklen, Zelle_5um_168h_VZ_5_C10_Lücke_Zyklen, Zelle_5um_168h_VZ_6_C10_Lücke_Zyklen])


# Berechnung des SOH für die Zellen mit 10 µm Li-Folie C/2-Zyklen
Zellen_10um_1h_C2_Mittlere_Spezifische_Kapazität_SOH,  Zellen_10um_1h_C2_Std_SOH = SOH_Berechnen_C2_C10([Zelle_10um_1h_VZ_1_C2_Lücke_Zyklen, Zelle_10um_1h_VZ_2_C2_Lücke_Zyklen, Zelle_10um_1h_VZ_3_C2_Lücke_Zyklen, Zelle_10um_1h_VZ_4_C2_Lücke_Zyklen, Zelle_10um_1h_VZ_5_C2_Lücke_Zyklen, Zelle_10um_1h_VZ_6_C2_Lücke_Zyklen])
# Berechnung des SOH für die Zellen mit 10 µm Li-Folie C/10-Zyklen
Zellen_10um_1h_C10_Mittlere_Spezifische_Kapazität_SOH,  Zellen_10um_1h_C10_Std_SOH = SOH_Berechnen_C2_C10([Zelle_10um_1h_VZ_1_C10_Lücke_Zyklen, Zelle_10um_1h_VZ_2_C10_Lücke_Zyklen, Zelle_10um_1h_VZ_3_C10_Lücke_Zyklen, Zelle_10um_1h_VZ_4_C10_Lücke_Zyklen, Zelle_10um_1h_VZ_5_C10_Lücke_Zyklen, Zelle_10um_1h_VZ_6_C10_Lücke_Zyklen])

# Berechnung des SOH für die Zellen mit 10 µm Li-Folie C/2-Zyklen
Zellen_10um_24h_C2_Mittlere_Spezifische_Kapazität_SOH,  Zellen_10um_24h_C2_Std_SOH = SOH_Berechnen_C2_C10([Zelle_10um_24h_VZ_1_C2_Lücke_Zyklen, Zelle_10um_24h_VZ_2_C2_Lücke_Zyklen, Zelle_10um_24h_VZ_3_C2_Lücke_Zyklen])
# Berechnung des SOH für die Zellen mit 10 µm Li-Folie C/10-Zyklen
Zellen_10um_24h_C10_Mittlere_Spezifische_Kapazität_SOH,  Zellen_10um_24h_C10_Std_SOH = SOH_Berechnen_C2_C10([Zelle_10um_24h_VZ_1_C10_Lücke_Zyklen, Zelle_10um_24h_VZ_2_C10_Lücke_Zyklen, Zelle_10um_24h_VZ_3_C10_Lücke_Zyklen])

# Berechnung des SOH für die Zellen mit 10 µm Li-Folie C/2-Zyklen
Zellen_10um_48h_C2_Mittlere_Spezifische_Kapazität_SOH,  Zellen_10um_48h_C2_Std_SOH = SOH_Berechnen_C2_C10([Zelle_10um_48h_VZ_1_C2_Lücke_Zyklen, Zelle_10um_48h_VZ_2_C2_Lücke_Zyklen, Zelle_10um_48h_VZ_3_C2_Lücke_Zyklen, Zelle_10um_48h_VZ_4_C2_Lücke_Zyklen])
# Berechnung des SOH für die Zellen mit 10 µm Li-Folie C/10-Zyklen
Zellen_10um_48h_C10_Mittlere_Spezifische_Kapazität_SOH,  Zellen_10um_48h_C10_Std_SOH = SOH_Berechnen_C2_C10([Zelle_10um_48h_VZ_1_C10_Lücke_Zyklen, Zelle_10um_48h_VZ_2_C10_Lücke_Zyklen, Zelle_10um_48h_VZ_3_C10_Lücke_Zyklen, Zelle_10um_48h_VZ_4_C10_Lücke_Zyklen])

# Berechnung des SOH für die Zellen mit 10 µm Li-Folie C/2-Zyklen
Zellen_10um_168h_C2_Mittlere_Spezifische_Kapazität_SOH,  Zellen_10um_168h_C2_Std_SOH = SOH_Berechnen_C2_C10([Zelle_10um_168h_VZ_1_C2_Lücke_Zyklen, Zelle_10um_168h_VZ_2_C2_Lücke_Zyklen, Zelle_10um_168h_VZ_3_C2_Lücke_Zyklen, Zelle_10um_168h_VZ_4_C2_Lücke_Zyklen, Zelle_10um_168h_VZ_5_C2_Lücke_Zyklen, Zelle_10um_168h_VZ_6_C2_Lücke_Zyklen])
# Berechnung des SOH für die Zellen mit 10 µm Li-Folie C/10-Zyklen
Zellen_10um_168h_C10_Mittlere_Spezifische_Kapazität_SOH,  Zellen_10um_168h_C10_Std_SOH = SOH_Berechnen_C2_C10([Zelle_10um_168h_VZ_1_C10_Lücke_Zyklen, Zelle_10um_168h_VZ_2_C10_Lücke_Zyklen, Zelle_10um_168h_VZ_3_C10_Lücke_Zyklen, Zelle_10um_168h_VZ_4_C10_Lücke_Zyklen, Zelle_10um_168h_VZ_5_C10_Lücke_Zyklen, Zelle_10um_168h_VZ_6_C10_Lücke_Zyklen])
```
